```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

image = cv2.imread(image_path)

print("Image uploaded successfully!")
print("Image dimensions:", image.shape)
```

```python

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 6))
plt.imshow(gray_image, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python

f_transform = np.fft.fft2(gray_image)

f_shift = np.fft.fftshift(f_transform)

magnitude_spectrum = 20 * np.log(np.abs(f_shift) + 1)
```

```python

plt.figure(figsize=(8, 6))
plt.imshow(magnitude_spectrum, cmap='gray')
plt.title("Fourier Transform Magnitude Spectrum")
plt.axis("off")
plt.show()
```

```python

uploaded_video = files.upload()

video_path = list(uploaded_video.keys())[0]

print("Video uploaded successfully!")
print("Video name:", video_path)
```

```python

cap = cv2.VideoCapture(video_path)

if cap.isOpened():
    print("Video opened successfully!")
else:
    print("Error opening video.")
```

```python

ret, frame = cap.read()

if ret:
    print("Video frame extracted successfully!")
else:
    print("Could not extract video frame.")
```

```python

frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(frame_rgb)
plt.title("Extracted Video Frame")
plt.axis("off")
plt.show()
```

```python

gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 6))
plt.imshow(gray_frame, cmap='gray')
plt.title("Grayscale Video Frame")
plt.axis("off")
plt.show()
```

```python

video_fft = np.fft.fft2(gray_frame)

video_fft_shift = np.fft.fftshift(video_fft)

video_magnitude = 20 * np.log(np.abs(video_fft_shift) + 1)
```

```python

plt.figure(figsize=(8, 6))
plt.imshow(video_magnitude, cmap='gray')
plt.title("Fourier Transform of Video Frame")
plt.axis("off")
plt.show()
```
